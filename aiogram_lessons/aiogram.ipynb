```python
import os
from time import sleep

import requests
import json
from load_dotenv import load_dotenv

load_dotenv()
token = os.getenv('TOKEN')
api_url = os.getenv('API_URL')
```

```python
method = 'getMe'
responce = requests.get(f'https://api.telegram.org/bot{token}/{method}')
print(responce.status_code)
print(responce.text)
```

```python
method = 'getUpdates'
responce = requests.get(f'https://api.telegram.org/bot{token}/{method}')
print(responce.status_code)
print(responce.text)
data = json.loads(responce.text)
chat_id = data['result'][0]['message']['from']['id']
name = data['result'][0]['message']['from']['first_name']
print(chat_id, name)
```

```python
"""send messsage"""
responce = requests.get(f'https://api.telegram.org/bot{token}/sendMessage?chat_id={chat_id}&text={"Hello, " + name}')
print(responce.status_code)
print(responce.text)
```

```python
"""эхо бот"""
MAX_COUNT = 200
COUNTER = 1
OFFSET = -1
method = 'getUpdates?offset='
while COUNTER < MAX_COUNT:
    print(f'Attempt {COUNTER}')
    data = requests.get(f'{api_url}{token}/{method}{OFFSET}').json()
    if data['result']:
        for update in data['result']:
            chat_id = update['message']['from']['id']
            name = update['message']['from']['first_name']
            requests.get(f'{api_url}{token}/sendMessage?chat_id={chat_id}&text={"Hello, " + name}')
            OFFSET = update['update_id'] + 1
    COUNTER += 1
    sleep(1)
```

```python
"""Вторая апишка с котами"""
MAX_COUNT = 20
COUNTER = 1
OFFSET = -1
method = 'getUpdates?offset='
while COUNTER < MAX_COUNT:
    print(f'Attempt {COUNTER}')
    data = requests.get(f'{api_url}{token}/{method}{OFFSET}').json()
    if data['result']:
        for update in data['result']:
            chat_id = update['message']['from']['id']
            name = update['message']['from']['first_name']
            requests.get(f'{api_url}{token}/sendMessage?chat_id={chat_id}&text={"Hello, " + name}')
            random_cat = requests.get('https://api.thecatapi.com/v1/images/search')
            if random_cat.status_code == 200:
                photo_url = random_cat.json()[0]['url']
                requests.get(f'{api_url}{token}/sendPhoto?chat_id={chat_id}&photo={photo_url}')
            else:
                requests.get(f'{api_url}{token}/sendMessage?chat_id={chat_id}&text=Здесь должен был быть кот...')

            OFFSET = update['update_id'] + 1
    COUNTER += 1
    sleep(1)
```

```python
"""long polling (timeout)"""
MAX_COUNT = 200
COUNTER = 1
timeout = 50
OFFSET = -1
method = 'getUpdates?offset='
while COUNTER < MAX_COUNT:
    print(f'Attempt {COUNTER}')
    data = requests.get(f'{api_url}{token}/{method}{OFFSET}&timeout={timeout}').json()
    if data['result']:
        for update in data['result']:
            chat_id = update['message']['from']['id']
            name = update['message']['from']['first_name']
            requests.get(f'{api_url}{token}/sendMessage?chat_id={chat_id}&text={"Hello, " + name}')
            OFFSET = update['update_id'] + 1
    COUNTER += 1
```

```python
import os
from aiogram import Bot, Dispatcher
from aiogram.filters import Command
from aiogram.types import Message
from load_dotenv import load_dotenv

```

```python
load_dotenv()
BOT_TOKEN = os.getenv('TOKEN')
bot = Bot(token=BOT_TOKEN)
dp = Dispatcher()


@dp.message(Command(commands='start'))
async def process_start_command(message: Message):
    await message.answer('Hello, write smth')


@dp.message(Command(commands='help'))
async def process_help_command(message: Message):
    await message.answer('Write smth, ill answer')


@dp.message()
async def send_echo(message: Message):
    await message.reply(text=message.text)


if __name__ == '__main__':
    dp.run_polling(bot)
```

```python
def custom_filter(some_list: list) -> bool:
    cnt = 0
    for item in some_list:
        if type(item) == int and item % 7 == 0:
            cnt += item
    return True if cnt <= 83 else False
```
